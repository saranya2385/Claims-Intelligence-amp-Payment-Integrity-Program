```python
import pandas as pd
from pathlib import Path

# 1. Set project folder
base_folder = Path.cwd()

# 2. Find all monthly CSV files
csv_files = []

for folder in base_folder.glob("Feed_20*"):
    if folder.is_dir():
        csv_files.extend(folder.rglob("*.csv"))

print("Total CSV files found:", len(csv_files))

# 3. Create output folder
output_folder = base_folder / "cleaned_data"
output_folder.mkdir(exist_ok=True)

# 4. Read, inspect, clean, and save each file
for file in csv_files:

    print("\nProcessing:", file.name)

    # Read CSV
    df = pd.read_csv(file)

    # Inspect original data
    print("Original shape:", df.shape)
    print("Missing values:", df.isnull().sum().sum())
    print("Duplicate rows:", df.duplicated().sum())

    # Clean column names
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
    )

    # Clean text columns
    text_cols = df.select_dtypes(
        include=["object", "string"]
    ).columns

    for col in text_cols:
        df[col] = df[col].str.strip()
        df[col] = df[col].replace("", pd.NA)

    # Remove exact duplicate rows
    before = len(df)
    df = df.drop_duplicates()
    removed = before - len(df)

    # Save cleaned file
    output_file = output_folder / (
        file.stem + "_cleaned.csv"
    )

    df.to_csv(output_file, index=False)

    # Display final results
    print("Cleaned shape:", df.shape)
    print("Duplicates removed:", removed)
    print("Remaining missing values:", df.isnull().sum().sum())
    print("Saved to:", output_file)

print("\nPipeline completed!")
```
